```python
import math

EMPTY = " "
MAX_PLAYER = "X"
MIN_PLAYER = "O"

nodes_minimax = 0
nodes_alphabeta = 0

# --------------------------------------------------
# Tampilan papan
# --------------------------------------------------
def print_board(board):
    print()

    for r in range(3):
        row = board[r * 3:(r + 1) * 3]
        print(" | ".join(row))

        if r < 2:
            print("--+---+--")

    print()

# --------------------------------------------------
# Pemeriksaan pemenang
# --------------------------------------------------
def winner(board):
    lines = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),

        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),

        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in lines:
        if (
            board[a] != EMPTY
            and board[a] == board[b]
            and board[b] == board[c]
        ):
            return board[a]

    return None

# --------------------------------------------------
# Terminal state
# --------------------------------------------------
def terminal(board):
    if winner(board) is not None:
        return True

    return EMPTY not in board

# --------------------------------------------------
# Utility
# --------------------------------------------------
def utility(board):
    w = winner(board)

    if w == MAX_PLAYER:
        return 10

    if w == MIN_PLAYER:
        return -10

    return 0

# --------------------------------------------------
# Daftar tindakan legal
# --------------------------------------------------
def actions(board):
    return [
        i
        for i in range(9)
        if board[i] == EMPTY
    ]

# --------------------------------------------------
# Menghasilkan keadaan baru
# --------------------------------------------------
def result(board, action, player):
    new_board = board.copy()

    new_board[action] = player

    return new_board

# --------------------------------------------------
# Fungsi evaluasi sederhana
# --------------------------------------------------
def evaluate(board):
    lines = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),

        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),

        (0, 4, 8),
        (2, 4, 6)
    ]

    score = 0

    for line in lines:
        values = [board[i] for i in line]

        if MIN_PLAYER not in values:
            score += values.count(MAX_PLAYER)

        if MAX_PLAYER not in values:
            score -= values.count(MIN_PLAYER)

    return score

# --------------------------------------------------
# Minimax
# --------------------------------------------------
def minimax(board, depth, maximizing):
    global nodes_minimax

    nodes_minimax += 1

    if terminal(board):
        return utility(board)

    if depth == 0:
        return evaluate(board)

    if maximizing:
        best = -math.inf

        for action in actions(board):
            child = result(
                board,
                action,
                MAX_PLAYER
            )

            value = minimax(
                child,
                depth - 1,
                False
            )

            best = max(best, value)

        return best

    else:
        best = math.inf

        for action in actions(board):
            child = result(
                board,
                action,
                MIN_PLAYER
            )

            value = minimax(
                child,
                depth - 1,
                True
            )

            best = min(best, value)

        return best

# --------------------------------------------------
# Menentukan langkah terbaik Minimax
# --------------------------------------------------
def best_move_minimax(board, depth):
    best_value = -math.inf
    best_action = None

    for action in actions(board):
        child = result(
            board,
            action,
            MAX_PLAYER
        )

        value = minimax(
            child,
            depth - 1,
            False
        )

        if value > best_value:

            best_value = value
            best_action = action

    return best_action, best_value

# --------------------------------------------------
# Alpha-Beta
# --------------------------------------------------
def alphabeta(
    board,
    depth,
    alpha,
    beta,
    maximizing
):
    global nodes_alphabeta

    nodes_alphabeta += 1

    if terminal(board):
        return utility(board)

    if depth == 0:
        return evaluate(board)

    if maximizing:
        value = -math.inf

        for action in actions(board):
            child = result(
                board,
                action,
                MAX_PLAYER
            )

            value = max(
                value,
                alphabeta(
                    child,
                    depth - 1,
                    alpha,
                    beta,
                    False
                )
            )

            alpha = max(alpha, value)

            if alpha >= beta:
                break

        return value

    else:
        value = math.inf

        for action in actions(board):
            child = result(
                board,
                action,
                MIN_PLAYER
            )

            value = min(
                value,
                alphabeta(
                    child,
                    depth - 1,
                    alpha,
                    beta,
                    True
                )
            )

            beta = min(beta, value)

            if alpha >= beta:
                break

        return value

# --------------------------------------------------
# Langkah terbaik Alpha-Beta
# --------------------------------------------------
def best_move_alphabeta(board, depth):
    best_value = -math.inf
    best_action = None

    alpha = -math.inf
    beta = math.inf

    for action in actions(board):
        child = result(
            board,
            action,
            MAX_PLAYER
        )

        value = alphabeta(
            child,
            depth - 1,
            alpha,
            beta,
            False
        )

        if value > best_value:
            best_value = value
            best_action = action

        alpha = max(alpha, best_value)

    return best_action, best_value

# --------------------------------------------------
# Program utama
# --------------------------------------------------
if __name__ == "__main__":
    board = [
        "X", "O", "X",
        " ", "O", " ",
        " ", " ", " "
    ]

    print("Keadaan awal:")
    print_board(board)

    depth = 6

    nodes_minimax = 0

    move, value = best_move_minimax(
        board,
        depth
    )

    print("MINIMAX")
    print("Langkah terbaik :", move)
    print("Nilai           :", value)
    print("Node dievaluasi :", nodes_minimax)

    nodes_alphabeta = 0

    move, value = best_move_alphabeta(
        board,
        depth
    )

    print("\nALPHA-BETA")
    print("Langkah terbaik :", move)
    print("Nilai           :", value)
    print("Node dievaluasi :", nodes_alphabeta)
```
